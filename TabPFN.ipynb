```python
df = pd.read_pickle("screw_data_s02-v2_identical-to-v1.pkl")
df.head()
```

```python
import numpy as np
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, StratifiedKFold, cross_val_score
from sklearn.metrics import make_scorer, f1_score

x_data = np.array(df['torque_values'].tolist())
y_data = np.array(df['class_values'].tolist())

le = LabelEncoder()
y_encoded = le.fit_transform(y_data)

X_train, X_test, y_train, y_test = train_test_split(x_data, y_encoded, test_size=0.2, stratify=y_encoded, random_state=42)
print(f"Train shape: {X_train.shape}, Test shape: {X_test.shape}")
model = TabPFNClassifier()
cv = StratifiedKFold(n_splits=3, shuffle=True,random_state=42)

f1_macro = make_scorer(f1_score, average="macro")

scores = cross_val_score(model,X_train,y_train,cv=cv,scoring=f1_macro,n_jobs=1)

print("TabPFN 3fold cvscores:", scores)
print("meanF1 Macro:", np.mean(scores))
print("F1 std:", np.std(scores))
```

```python
final_model = TabPFNClassifier()
final_model.fit(X_train, y_train)
y_pred = final_model.predict(X_test)
test_f1_macro = f1_score(y_test, y_pred, average="macro")
print("test f1 macro:", test_f1_macro)
```
