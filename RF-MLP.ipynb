## Random Forest Reproduction seed split index based

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.utils import shuffle
import numpy as np

x_data = np.array(df['torque_values'].tolist())
y_data = np.array(df['class_values'].tolist())

le = LabelEncoder()
y_encoded = le.fit_transform(y_data)

split_ratio = 0.65
n_samples = len(x_data)
split_idx = int(n_samples * split_ratio)

X_s, y_s = shuffle(x_data, y_encoded)

x_train = X_s[:split_idx,]
y_train = y_s[:split_idx,]
x_test = X_s[split_idx:]
y_test = y_s[split_idx:]
```

```python
from sklearn.ensemble import RandomForestClassifier
rf = RandomForestClassifier(
    n_estimators=500,
    random_state=42,
    n_jobs=-1
)

rf.fit(x_train, y_train)
y_pred = rf.predict(x_test)
```

```python
from sklearn.metrics import f1_score
f1 = f1_score(y_test, y_pred, average="macro")
print("Macro F1:", f1)
```

## MLPClassifier seed split index based

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.metrics import f1_score
from sklearn.utils import shuffle
from sklearn.neural_network import MLPClassifier
import numpy as np
x_data = np.array(df['torque_values'].tolist())
y_data = np.array(df['class_values'].tolist())
le = LabelEncoder()
y_encoded = le.fit_transform(y_data)
seeds = [1, 42, 56, 73, 81]

split_ratio = 0.65
n_samples = len(x_data)
split_idx = int(n_samples * split_ratio)
f1_scores = []
```

```python
for s in seeds:
    X_s, y_s = shuffle(x_data,y_encoded,random_state=s)
    x_train = X_s[:split_idx]
    y_train = y_s[:split_idx]

    x_test = X_s[split_idx:]
    y_test = y_s[split_idx:]
    mlp = MLPClassifier(hidden_layer_sizes=(128,64),activation="relu", solver="adam", max_iter=500,random_state=s)

    mlp.fit(x_train, y_train)
    y_pred = mlp.predict(x_test)
    f1 = f1_score( y_test,y_pred,average="macro")

    f1_scores.append(f1)

    print(f"Seed {s} - macro F1: {f1:.4f}")


mean_f1 = np.mean(f1_scores)
std_f1 = np.std(f1_scores)
print(f"mean macro F1: {mean_f1:.4f}")
print(f"Std macro F1: {std_f1:.4f}")
```

```python
from sklearn.neural_network import MLPClassifier
mlp = MLPClassifier(hidden_layer_sizes=(128,64),activation="relu", solver="adam", max_iter=500,random_state=s)
mlp.fit(x_train, y_train)
y_pred_mlp = mlp.predict(x_test)
f1 = f1_score(y_test, y_pred_mlp, average="macro")
print("Macro F1:", f1)
```

## MLP Classifier 3-CV Fold

```python
from sklearn.preprocessing import LabelEncoder
from sklearn.model_selection import train_test_split, StratifiedKFold
from sklearn.neural_network import MLPClassifier
from sklearn.metrics import f1_score
import numpy as np
x_data = np.array(df['torque_values'].tolist())
y_data = np.array(df['class_values'].tolist())

le = LabelEncoder()
y_encoded = le.fit_transform(y_data)

x_train, x_test, y_train, y_test = train_test_split(x_data,y_encoded,test_size=0.2,random_state=42, stratify=y_encoded)

skf = StratifiedKFold(n_splits=3, shuffle=True, random_state=81)

cv_scores_val = []
best_val_f1 = -1
best_model = None

for fold, (train_index, val_index) in enumerate(skf.split(x_train, y_train)):
    x_train_fold, x_val_fold = x_train[train_index], x_train[val_index]
    y_train_fold, y_val_fold = y_train[train_index], y_train[val_index]
    print(f"Fold{fold+1}")

    mlp = MLPClassifier(hidden_layer_sizes=(128, 64), activation="relu",solver="adam",max_iter=500,random_state=42)
    mlp.fit(x_train_fold, y_train_fold)
    y_val_pred = mlp.predict(x_val_fold)
    val_f1 = f1_score(y_val_fold, y_val_pred, average="macro")
    cv_scores_val.append(val_f1)
    print(f"Validation Macro F1: {val_f1:.4f} im Fold {fold+1}")

    if val_f1 > best_val_f1:
        best_val_f1 = val_f1
        best_model = mlp
    
    
print(f"CV-Ergebnis - Validation Macro F1: {np.mean(cv_scores_val):.4f} ± {np.std(cv_scores_val):.4f}")
test_pred = best_model.predict(x_test)
test_f1 = f1_score(y_test, test_pred, average="macro")
print(f"test macro f1: {test_f1:.4f}")
    
        

```
